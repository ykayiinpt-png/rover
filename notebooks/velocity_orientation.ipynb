```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import ScalarFormatter
```

```python
# =========================
# Lecture du fichier CSV
# =========================
csv_file = ".\\inputs\\pid_orientation_right_wheel.csv"
```

```python
df = pd.read_csv(csv_file)

# =========================
# Garder à partir de la 300ème valeur
# =========================
df = df.iloc[3200:3750].reset_index(drop=True)


# =========================
# Conversion du temps
# =========================
# Temps relatif en secondes
t = df["Command_x"] - df["Command_x"].iloc[0]

# =========================
# Style scientifique
# =========================
plt.style.use("seaborn-v0_8-whitegrid")

fig, ax = plt.subplots(figsize=(10, 5), dpi=150)

# =========================
# Tracé des courbes
# =========================
ax.plot(
    t,
    df["Command_y"],
    label="Command",
    linewidth=2,
    color="tab:blue"
)

ax.plot(
    t,
    df["Target_y"],
    label="Target",
    linewidth=2,
    linestyle="--",
    color="tab:red"
)

# =========================
# Mise en forme scientifique
# =========================
ax.set_title("", fontsize=16, fontweight="bold")
ax.set_xlabel("Temps [s]", fontsize=13)
ax.set_ylabel("Vitesse (RPM)", fontsize=13)

# Affichage scientifique des axes
formatter = ScalarFormatter(useMathText=True)
formatter.set_scientific(True)
formatter.set_powerlimits((-3, 3))

ax.xaxis.set_major_formatter(formatter)
ax.yaxis.set_major_formatter(formatter)

# Grille scientifique
ax.grid(True, which='major', linestyle='--', alpha=0.7)
ax.minorticks_on()
ax.grid(True, which='minor', linestyle=':', alpha=0.4)

# Légende
ax.legend(fontsize=11)

# Ajustement automatique
fig.tight_layout()

# =========================
# Affichage
# =========================
plt.show()
```

```python
# =========================
# Axe X = index des échantillons
# =========================
from matplotlib.ticker import MultipleLocator


x = df.index

# =========================
# Style scientifique
# =========================
plt.style.use("seaborn-v0_8-whitegrid")

fig, ax = plt.subplots(figsize=(10, 5), dpi=150)

# =========================
# Tracé des données
# =========================
ax.plot(
    x,
    df["Command_y"],
    label="Command",
    linewidth=1.8,
    color="tab:blue"
)

ax.plot(
    x,
    df["Target_y"],
    label="Target",
    linewidth=1.8,
    linestyle="--",
    color="tab:red"
)

# =========================
# Mise en forme
# =========================
ax.set_title("Command vs Target", fontsize=16, fontweight="bold")

ax.set_xlabel("Index des échantillons", fontsize=13)
ax.set_ylabel("Amplitude", fontsize=13)

# Graduation tous les 100 points
ax.xaxis.set_major_locator(MultipleLocator(100))

# Grilles
ax.grid(True, which='major', linestyle='--', alpha=0.7)
ax.minorticks_on()
ax.grid(True, which='minor', linestyle=':', alpha=0.4)

# Légende
ax.legend(fontsize=11)

# Ajustement automatique
fig.tight_layout()

# =========================
# Affichage
# =========================
plt.show()
```
