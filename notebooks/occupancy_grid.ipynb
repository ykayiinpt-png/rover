```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load the grid from the .npy file
file_path = '.\\inputs\\occupancy_grid_1778706273.659889.npy'  # Replace with your .npy file path
grid = np.load(file_path)
```

```python
grid
```

```python
# Check the shape
print(f"Grid shape: {grid.shape}")
print(f"Grid min/max values: {grid.min()}, {grid.max()}")
```

```python
grid = 1 - 1 / (1 + np.exp(grid))
```

```python
grid
```

```python
fig, ax = plt.subplots()

ax.imshow(grid, cmap="gray", origin="lower")

cx = grid.shape[1] // 2
cy = grid.shape[0] // 2

ax.plot(cx, cy, "ro", markersize=6)

ax.set_xticks(range(grid.shape[1]))
ax.set_yticks(range(grid.shape[0]))

ax.set_xticks([x - 0.5 for x in range(grid.shape[1])], minor=True)
ax.set_yticks([y - 0.5 for y in range(grid.shape[0])], minor=True)

ax.grid(which="minor", color="black", linewidth=0.3)

plt.show()
```

```python
# Display as a colormap
plt.grid(True, color="red", linewidth=0.1)
plt.xticks(range(grid.shape[1]))
plt.yticks(range(grid.shape[0]))
plt.imshow(grid, cmap='gray_r')  # You can choose other colormaps like 'hot', 'cool', 'plasma'
plt.colorbar()  # Show the color scale
plt.title('2D Grid Visualization')
plt.show()
```
