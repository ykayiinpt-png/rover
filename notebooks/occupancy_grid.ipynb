```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load the grid from the .npy file
file_path = '.\\inputs\\occupancy_grid_1778631623.203515.npy'  # Replace with your .npy file path
grid = np.load(file_path)
```

```python
# Check the shape
print(f"Grid shape: {grid.shape}")
print(f"Grid min/max values: {grid.min()}, {grid.max()}")
```

```python
# Display as a colormap
plt.imshow(grid, cmap='viridis')  # You can choose other colormaps like 'hot', 'cool', 'plasma'
plt.colorbar()  # Show the color scale
plt.title('2D Grid Visualization')
plt.show()
```
